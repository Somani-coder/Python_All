```python
# arbitary arguments

#you can pass a variable number of arguments to a function using *args & **kwargs
#using *args for non keyword arguments
```

```python
def add(*args):
    return sum(args)
print(add(1,2,3))
print(add(1,2,3,5,6,7))
```

```python
# Using **kwargs for Keyword Arguments

# **kwargs: Collects extra keyword arguments as a dictionary.
# kwargs.items(): Iterates over key-value pairs in the dictionary.

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="Alice", age=30, city="New York")
```

```python
#lambda functions
#these are small anonymous functions defined with lambda kewword
#when we want to store some functions for the temporary time, we use lambda functions
```

```python
#normal function
def add (a,b):
    return a + b
print("addition =",add(8,5))

#lambda function
add = lambda a,b:a+b
print("L_addition =",add(8,5))
```

```python
#addition
add = lambda a,b:a+b
print("L_addition =",add(8,5))

#subtraction
sub = lambda a,b:a-b
print("L_subtraction =",sub(8,5))

#multiplication
mul = lambda a,b:a*b
print("L_multiplication =",mul(8,5))

#division
div = lambda a,b:a/b
print("L_division =",div(8,5))

#find a square of a number
sqr = lambda a: a*a
print("square =",sqr(25))
```
