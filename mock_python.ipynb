```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('titanic.csv')
```

```python
# How many rows and columns are in the dataset?
```

```python
df.shape
```

```python
# What are the column names and data types?
```

```python
df.dtypes
```

```python
# What is the average age of the passengers?
```

```python
avg_age = df['age'].mean()
print(avg_age)
```

```python
# How many passengers survived?
```

```python
survived_count = df['survived'].sum()
print(survived_count)
```

```python
# How many males and females are in the dataset?
```

```python
gender_counts=df['sex'].value_counts()
print(gender_counts)
```

```python
# What is the most common embarkation point?
```

```python
common_embarkation = df['embark_town'].mode()
print(common_embarkation)
```

```python
# Create a bar chart showing the number of male vs female passengers.
```

```python
gender_counts = df['sex'].value_counts()
sns.barplot(x=gender_counts.index, y=gender_counts.values, color='crimson')
plt.xlabel('Sex')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
# Create a histogram showing the distribution of passenger ages.
```

```python
plt.hist(df['age'], edgecolor='black',color='yellow')
plt.xlabel("Age")
plt.ylabel("No of Passengers")
plt.show()
```

```python
# Create a pie chart of survival (Survived vs Not Survived).
```

```python
survival_counts = df['survived'].value_counts()
labels = ['not survived', 'survived']
plt.pie(survival_counts, labels=labels, autopct='%1.1f%%', startangle=90, colors=['grey', 'pink'])
plt.title("survival Distribution")
plt.axis('equal')
plt.show()
```

```python
# Create a bar chart showing the number of passengers in each class (Pclass).
```

```python
class_counts = df['pclass'].value_counts()
sns.barplot(x=class_counts.index, y=class_counts.values, palette='muted')
plt.title('Number of Passengers in Each Class')
plt.xlabel('Passenger Class (Pclass)')
plt.ylabel('Number of Passengers')
plt.show()
```
