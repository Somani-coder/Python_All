```python
#join and concatenation in pandas
```

```python
#concatenation along rows(axis 0)
df1 = pd.DataFrame({
    'A': ['A0','A1','A2'],
    'B': ['B0','B1','B2'] 
})
print(df1)

df2 = pd.DataFrame({
    'A': ['A3','A4','A5'],
    'B': ['B3','B4','B5'] 
})
print(df2)

result =pd.concat([df1,df2])
print(f"concat df1 & df2\n: {result}")

#here we have not mentioned the axis so by default
#the result will be row wise
```

```python
#concatenate along columns (axis 1)
result = pd.concat([df1,df2],axis=1)
print(result)
```

```python
#for continuous indexing
```

```python
df1 = pd.DataFrame({
    'A': ['A0','A1','A2'],
    'B': ['B0','B1','B2'] 
})
print(df1)

df2 = pd.DataFrame({
    'A': ['A3','A4','A5'],
    'B': ['B3','B4','B5'] 
})
print(df2)

result =pd.concat([df1,df2], ignore_index=True)
print(f"concat df1 & df2\n: {result}")
```

```python
#merge
```

```python
#inner merge
#returns only the rows where the key exists in both DataFrames
```

```python
df1 = pd.DataFrame({
    'key': ['A','B','C','D'],
    'value': [1,2,3,4]
})
print(df1)
df2 = pd.DataFrame({
    'key': ['B','D','E','F'],
    'value': [5,6,7,8]
})
print(df2)
result = pd.merge(df1,df2,on='key',how='inner')
print(result)
```
