```python
import array as arr

#Creating an Array
#You create an array by specifying the type code and the initial values:

Eg_array = arr.array('i', [1, 2, 3, 4, 5])

#'i': Type code for integers.
#[1, 2, 3, 4, 5]: Initial values for the array.

#You can access array elements by index:
print(Eg_array[0])     
print(Eg_array[2])     
print(Eg_array[3])     
```

```python
import array as x
x = x.array('i',[1,2,3,4,5,6,26,5])
print(x)

#append()

x.append(8)
print(x)
```

```python
#extend()

x.extend([12,14,14])
print(x)
```

```python
#insert()

x.insert(4,24)      #insert 24 in the 4th index
print(x)
```

```python
#pop()
x.pop()      #remove and returns the last element
print(x)
```

```python
x.pop(2)
print(x)   #removes and returns the number presnt in the 2nd index i.e 3
```

```python
#remove()
import array as x
x = x.array('i',[1,2,3,4,5,6,26,5])
x.remove(5) #remove() will remove the first occurance ,another 5 is present
print(x)
```

```python
#count()
import array as arr
arr = arr.array('i',[1,2,3,4,5,6,26,5,4,4,4])

print("original array:", arr)
print(arr.count(4))
```

```python
#reverse()
import array as x
x = x.array('i',[1,2,3,4,5,6,26,5])
print("original array: ", x)
x.reverse()
print("reversed array: ", x)
```
